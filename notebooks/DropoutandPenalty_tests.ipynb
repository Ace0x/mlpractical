```python
# Diagnostic: Test dropout scaling directly
import numpy as np
from mlp.layers import DropoutLayer

print("Testing dropout layer scaling...")
rng_test = np.random.RandomState(12345)
layer_test = DropoutLayer(rng=rng_test, incl_prob=0.5)
x_test = np.array([[1.0, 2.0, 3.0, 4.0]])
out_test = layer_test.fprop(x_test)

print(f"Input: {x_test}")
print(f"Output: {out_test}")
non_zero = out_test[out_test != 0]
if len(non_zero) > 0:
    original_val = x_test.flat[np.where(out_test.flat != 0)[0][0]]
    print(f"Scaling factor: {non_zero[0] / original_val} (should be 2.0 for inverted dropout)")
else:
    print("All values were dropped (try again with different seed)")
    
print(f"\nincl_prob = {layer_test.incl_prob}")
```

```python
# 1) Is your mask shape right and identical to the grader's mask?
expected = fprop_correct
got = fprop_output

mask_expected = (expected != 0)
mask_got = (got != 0)

print("mask shape:", mask_got.shape, " expected:", mask_expected.shape)
print("mask equal? ->", np.array_equal(mask_got, mask_expected))
print("mask mismatch count:", np.sum(mask_got ^ mask_expected))

# 2) If masks match but values don’t, your scaling is off.
#    Check the ratio at kept positions (should be ~1)
kept = mask_expected & mask_got
if np.any(kept):
    ratio = got[kept] / expected[kept]
    print("median ratio at kept positions:", np.median(ratio))
    print("unique ratios (approx):", np.unique(np.round(ratio, 3))[:10])

```
